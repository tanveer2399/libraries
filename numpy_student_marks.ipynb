# NumPy Student Marks Analysis

Use NumPy to calculate each student's total and average marks, the average for each subject, and identify the student with the highest total.

```python
import numpy as np

marks = np.array([
    [80, 75, 90, 85, 70],
    [60, 65, 55, 70, 68],
    [95, 92, 88, 90, 96],
    [39, 50, 55, 60, 48]
])

marks
```

```python
# Rows represent students, so axis=1 calculates across each student's subjects.
total_marks = np.sum(marks, axis=1)
average_marks_per_student = np.mean(marks, axis=1)

print("Total marks for each student:", total_marks)
print("Average marks for each student:", average_marks_per_student)
```

```python
# Columns represent subjects, so axis=0 calculates down each subject column.
average_marks_per_subject = np.mean(marks, axis=0)
print("Average marks for each subject:", average_marks_per_subject)
```

```python
highest_student_index = np.argmax(total_marks)
print("Student with the highest total (zero-based index):", highest_student_index)
print("Highest total marks:", total_marks[highest_student_index])
```
